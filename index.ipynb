## Module 1: Functional Interfaces & Lambdas
Java 8 introduced functional interfaces (interfaces with one abstract method). 
Let's try the **Predicate** (takes a value, returns a boolean).

```java
import java.util.function.Predicate;
import java.util.List;

List<Integer> numbers = List.of(1, 2, 3, 4, 5, 6);
Predicate<Integer> isEven = n -> n % 2 == 0;

System.out.println("Filtering even numbers:");
numbers.stream().filter(isEven).forEach(System.out::println);
```

## Module 2: The Stream API
Streams allow us to process data pipelines. `map` transforms data, while `flatMap` flattens nested lists.

```java
import java.util.stream.Collectors;

List<List<String>> nested = List.of(List.of("Java", "Spring"), List.of("Docker", "K8s"));

List<String> flattened = nested.stream()
    .flatMap(List::stream)
    .map(String::toUpperCase)
    .collect(Collectors.toList());

System.out.println("Flattened and Uppercased: " + flattened);
```

## Module 3: Modern Data Modeling (Records)
Records are immutable data carriers introduced to reduce boilerplate code.

```java
public record Developer(String name, String language) {}

Developer dev = new Developer("Alex", "Java");
System.out.println("Record auto-generated toString: " + dev);
System.out.println("Accessing field (no get prefix): " + dev.name());
```

## Module 4: Virtual Threads (Java 21)
Virtual threads are lightweight threads managed by the JVM. Let's start one!

```java
Thread vThread = Thread.startVirtualThread(() -> {
    System.out.println("Running inside a Virtual Thread: " + Thread.currentThread());
});

vThread.join(); // Wait for it to finish
```
