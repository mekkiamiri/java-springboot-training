## Module 1: Functional Interfaces & Lambdas
A functional interface has exactly one abstract method.

```java
import java.util.function.Predicate;

// Lambda Expression
Predicate<String> isLongLambda = s -> s.length() > 10;
System.out.println("Is 'introduction' a long string? " + isLongLambda.test("introduction"));
```

## Module 2: The Stream API
Grouping and Partitioning using Records.

```java
import java.util.List;
import java.util.Map;
import java.util.stream.Collectors;

public record Employee(String name, String department, int salary) {}

List<Employee> employees = List.of(
    new Employee("Alice", "Engineering", 120000),
    new Employee("Bob", "Engineering", 110000),
    new Employee("Charlie", "HR", 90000)
);

Map<String, List<Employee>> byDept = employees.stream()
    .collect(Collectors.groupingBy(Employee::department));

System.out.println("Departments: " + byDept.keySet());
```

## Module 3: Modern Data Modeling
Records with custom validation logic.

```java
public record PositiveAmount(int value) {
    public PositiveAmount {
        if (value < 0) throw new IllegalArgumentException("Negative amount!");
    }
}

try {
    new PositiveAmount(-10);
} catch (Exception e) {
    System.out.println("Caught expected error: " + e.getMessage());
}
```
